# Environment Variables Echo - Jupyter Notebook

## CONNECT Environment Variables

```python
import os
from datetime import datetime

# Get all environment variables starting with CONNECT_
connect_vars = {key: value for key, value in os.environ.items() if key.startswith('CONNECT_')}

if connect_vars:
    print("Environment variables starting with CONNECT_:\n")
    for key, value in sorted(connect_vars.items()):
        print(f"{key}={value}")
else:
    print("No environment variables starting with CONNECT_ found.")
```
